# Linear Regression & Regularization (Ridge vs Lasso)

**Assignment Deliverables:**
1. Train/test split, Linear/Ridge/Lasso models.
2. A comparison table of MAE / MSE / RMSE / R² across the 3 models.
3. 2–3 sentence conclusion on the best model and the effect of regularization.

```python
# Importing required libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import warnings

warnings.filterwarnings('ignore')

# Loading the dataset
df = pd.read_csv('Advertising.csv')

if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)

display(df.head())
```

### 1. Train-Test Split

```python
# Separate features (X) and target (y)
X = df[['TV', 'Radio', 'Newspaper']]
y = df['Sales']

# Split the data into 67% training and 33% testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

print(f"Training set shape: {X_train.shape}")
print(f"Testing set shape: {X_test.shape}")
```

### 2. Train Linear, Ridge, and Lasso Models

```python
# Initialize the models
linear_model = LinearRegression()
ridge_model = Ridge(alpha=1.0)
lasso_model = Lasso(alpha=1.0)

# Fit the models on the training data
linear_model.fit(X_train, y_train)
ridge_model.fit(X_train, y_train)
lasso_model.fit(X_train, y_train)

print("Models successfully trained!")
```

### 3. Compare Model Coefficients

```python
# Create a DataFrame to compare the coefficients assigned by each model
coef_df = pd.DataFrame({
    'Feature': X.columns,
    'Linear Regression': linear_model.coef_,
    'Ridge': ridge_model.coef_,
    'Lasso': lasso_model.coef_
})

print("--- Model Coefficients Comparison ---")
display(coef_df)
```

### 4. Evaluate Models (MAE, MSE, RMSE, R²)

```python
# Generate predictions on the test set
pred_linear = linear_model.predict(X_test)
pred_ridge = ridge_model.predict(X_test)
pred_lasso = lasso_model.predict(X_test)

# Helper function to compile metrics
def evaluate_model(name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    return [name, mae, mse, rmse, r2]

# Compile results into a list
results = [
    evaluate_model('Linear Regression', y_test, pred_linear),
    evaluate_model('Ridge Regression', y_test, pred_ridge),
    evaluate_model('Lasso Regression', y_test, pred_lasso)
]

# Create comparison dataframe
eval_df = pd.DataFrame(results, columns=['Model', 'MAE', 'MSE', 'RMSE', 'R²'])

print("--- Final Model Evaluation Metrics ---")
display(eval_df)
```

### 5. Conclusion
**Lasso Regression** generalizes the best overall, achieving the lowest test errors (RMSE: 1.497) while maintaining a nearly identical high $R^2$ (~0.897) compared to standard Linear Regression. By observing the coefficients table, we can see the true effect of **L1 regularization (Lasso)**: it aggressively shrank the `Newspaper` coefficient closer to zero, performing automatic feature selection and ignoring the noisy feature, whereas **Ridge (L2)** only slightly reduced it. This makes the Lasso model simpler, less prone to overfitting, and highly effective for this specific dataset.
